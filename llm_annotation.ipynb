### Using OpenAI (GPT-5.4) for data annotation

```python
example_annotated = """1	Dated	O
2	March	O
3	31	O
4	,	O
5	2007	O
6	Thinkplus	I-ORG
7	Investments	I-ORG
8	Limited	I-ORG
9	(	O
10	as	O
11	the	O
12	Lender	I-PER
13	)	O
14	AND	O
15	Airland	I-ORG
16	International	I-ORG
17	Limited	I-ORG
18	Bizexpress	I-ORG
19	Limited	I-ORG
20	(	O
21	as	O
22	the	O
23	Borrower	I-PER
24	)	O
25	Loan	O
26	Agreement	O
27	Contents	O"""

example_annotated = example_annotated.split('\n')
example_annotated = [line.split('\t') for line in example_annotated]

tokens = []
labels = []

for line in example_annotated:
    tokens.append(line[1])
    labels.append(line[2])

print(tokens)
```

```python
from openai import OpenAI
client = OpenAI()

response = client.responses.create(
    model="gpt-5.4",  # example available model
    input = (
    "Label each token with BIO tags (ORG, PER, LOC, MISC).\n"
    "Borrower/lender = PER.\n\n"

    "Rules:\n"
    "- One label per token\n"
    "- Same order\n"
    "- Output only labels separated by spaces\n"
    "- Number of labels must match tokens\n\n"

    "Tokens:\n"
    + " ".join(tokens) +
    "\n\nOutput:"
    )
)

print(response.output[0].content[0].text)
```
