### Conclusions

Although ORG entities are infrequent in both datasets, the poor ORG performance when training only on synthetic data suggests that the issue is not only entity frequency but also limited diversity in the generated examples. While the synthetic dataset contains a reasonable density of ORG labels, many of the generated entities are repeated names appearing in very similar contexts, which likely limits the model’s ability to generalize to unseen ORG entities in FIN3. 

Although mixing synthetic and real in-domain data improves ORG detection compared to using synthetic data alone, the overall gains remain relatively modest considering the amount of synthetic data introduced. This may be explained by limited diversity in the generated LOC and ORG examples, where many entities appear repeatedly in similar contexts. More targeted prompting could therefore improve the effectiveness of the data augmentation strategy by encouraging a wider variety of LOC and ORG entities, as well as longer and more complex spans, which remain challenging for the model.
